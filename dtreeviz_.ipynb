```python
import dtreeviz
import graphviz.backend as be
```

```python
from sklearn import tree
from sklearn.datasets import *
from dtreeviz.trees import *
from IPython.display import Image, display_svg, SVG
```

```python
clas = tree.DecisionTreeClassifier()
iris = load_iris()

X_train = iris.data
y_train = iris.target
clas.fit(X_train, y_train)
```

```python
from sklearn.tree import plot_tree
plot_tree(clas)
```

# **1. Classification**

```python
import logging
logging.getLogger('matplotlib.font_manager').disabled = True

model_viz = dtreeviz.model(clas,
                         X_train=X_train,
                         y_train=y_train,
                         feature_names=iris.feature_names,
                         class_names=iris.target_names)
viz = model_viz.view(orientation='LR',scale=1)  # ORIENTATION --> LR/TD (TD by default)
viz
```

# **2. Regression**

```python
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()
regr = tree.DecisionTreeRegressor(max_depth=1)


X_train = housing.data
y_train = housing.target
regr.fit(X_train, y_train)

model_viz_regr = dtreeviz.model(regr,
               X_train=X_train,
               y_train=y_train,
               target_name='prices',
               feature_names=housing.feature_names
              )
viz = model_viz_regr.view(orientation='LR',scale=2.5)  # ORIENTATION --> LR/TD (TD by default)
viz
```

# **PREDICTION PATH**

```python
iris = load_iris()

X_train = iris.data
y_train = iris.target
clas.fit(X_train, y_train)

# pick one sample
X = iris.data[np.random.randint(0, len(iris.data)), :]

# build viz model
model_viz_pred = dtreeviz.model(
    clas,
    X_train=X_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    class_names=list(iris.target_names)
)

# highlight prediction path
viz = model_viz_pred.view(
    x=X,                # this highlights the path
    orientation="LR",
    scale=1
)

viz
```

```python
iris = load_iris()

X_train = iris.data
y_train = iris.target
clas.fit(X_train, y_train)

# pick one sample
X = iris.data[np.random.randint(0, len(iris.data)), :]

# build viz model
model_viz_pred = dtreeviz.model(
    clas,
    X_train=X_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    class_names=list(iris.target_names)
)

# highlight prediction path
viz = model_viz_pred.view(
    x=X,                # this highlights the path
    orientation="LR",
    scale=1,
    show_node_labels=True
)

viz
```
