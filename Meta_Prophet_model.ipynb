```python
from prophet import Prophet
```

```python
model=Prophet()
```

```python
df_f.columns
```

```python
df_f['da']=pd.to_datetime(df_f['da'], format='%d-%m-%Y')
df_f.head()
```

```python
df_f.rename(columns={'da': 'ds'}, inplace=True)
model.fit(df_f)
```

```python
future_dates=model.make_future_dataframe(periods=20)
future_dates.tail()
```

```python
pred=model.predict(future_dates)
```

```python
pred.head()
```

```python
model.plot(pred)
```

```python
model.plot_components(pred)
```

```python
# Now we have to do CROSS VALIDATION to check our model
# Lets assume we have data of 1960-1962
# cross validation will divide this dataset into 2 parts
# 1960-61 --> training dataset
# 1961-62 --> testing dataset
```

```python
from prophet.diagnostics import cross_validation
```

```python
df_cv=cross_validation(model,initial='15 days', period='1 day',horizon='3 days')
```

```python
df_cv.head()
```

```python
from prophet.diagnostics import performance_metrics
df_p=performance_metrics(df_cv)
df_p.head()
```
