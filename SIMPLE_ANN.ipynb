```python
import math
import random


def sigmoid(x):
    return 1 / (1 + math.exp(-x))


def initialize_network():
    network = {
        'input': [0.0, 0.0],
        'hidden1': [0.0, 0.0, 0.0],
        'hidden2': [0.0, 0.0, 0.0],
        'output': [0.0],


        'weights_input_hidden1': [[random.random() for _ in range(2)] for _ in range(3)],


        'weights_hidden1_hidden2': [[random.random() for _ in range(3)] for _ in range(3)],


        'weights_hidden2_output': [random.random() for _ in range(3)],


        'bias_hidden1': [random.random() for _ in range(3)],
        'bias_hidden2': [random.random() for _ in range(3)],
        'bias_output': random.random()
    }
    return network

def forward_pass(network, inputs):
    network['input'] = inputs

    for i in range(3):
        weighted_sum = sum(network['input'][j] * network['weights_input_hidden1'][i][j] for j in range(2)) + network['bias_hidden1'][i]
        network['hidden1'][i] = sigmoid(weighted_sum)


    for i in range(3):
        weighted_sum = sum(network['hidden1'][j] * network['weights_hidden1_hidden2'][i][j] for j in range(3)) + network['bias_hidden2'][i]
        network['hidden2'][i] = sigmoid(weighted_sum)


    weighted_sum = sum(network['hidden2'][i] * network['weights_hidden2_output'][i] for i in range(3)) + network['bias_output']
    network['output'][0] = sigmoid(weighted_sum)

    return network['output'][0]

test_data = [
    ([0, 0]),
    ([0, 1]),
    ([1, 0]),
    ([1, 1])
]


network = initialize_network()


print("Test Results (Forward Propagation):")
for inputs in test_data:
    output = forward_pass(network, inputs)
    print(f"Input: {inputs}, Predicted Output: {output}")
```
