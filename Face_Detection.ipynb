```python
from mtcnn import MTCNN
import cv2
```

```python
from google.colab.patches import cv2_imshow

detector =MTCNN()
img=cv2.imread("MS DHONI.jpg")
out=detector.detect_faces(img)
```

```python
out
```

```python
for i in out:
  x,y,w,h= i['box']

  left_eyeX, left_eyeY= i['keypoints']['left_eye']
  right_eyeX, right_eyeY= i['keypoints']['right_eye']
  noseX, noseY= i['keypoints']['nose']
  mouth_leftX, mouth_leftY= i['keypoints']['mouth_left']
  mouth_rightX, mouth_rightY= i['keypoints']['mouth_right']

  cv2.circle(img, (left_eyeX, left_eyeY), color=(255,0,0), radius=5, thickness=3)
  cv2.circle(img, (right_eyeX, right_eyeY), color=(255,0,0), radius=5, thickness=3)
  cv2.circle(img, (noseX, noseY), color=(255,0,0), radius=5, thickness=3)
  cv2.circle(img, (mouth_leftX, mouth_leftY), color=(255,0,0), radius=5, thickness=3)
  cv2.circle(img, (mouth_rightX, mouth_rightY), color=(255,0,0), radius=5, thickness=3)

  cv2.rectangle(img, (x,y), (x+w, y+h), color=(0,255,0), thickness=3)

cv2_imshow(img)
```
