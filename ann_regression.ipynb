```python
import pickle as pkl
import pandas as pd
import numpy as np
with open("CarPricesData.pkl", "rb") as f:
    object = pkl.load(f)

df = pd.DataFrame(object)
df.to_csv(r'file.csv')
```

```python
target=df['Price']
```

```python
predictor_var=['Age','KM','Weight','HP','MetColor','CC','Doors']
```

```python
x=df[predictor_var].values
print(x)
```

```python

y=target.values
print(y)
```

```python
from sklearn.preprocessing import StandardScaler
predicted_scalar=StandardScaler()
target_scalar=StandardScaler()
```

```python
predictor_scalar_fit=predicted_scalar.fit_transform(x)
target_scalar_fit=target_scalar.fit_transform(y.reshape(-1,1))
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(predictor_scalar_fit,target_scalar_fit,test_size=0.3,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from keras.models import Sequential
from keras.layers import Dense
```

```python
model=Sequential()
```

```python
model.add(Dense(units=5,input_dim=7, kernel_initializer='normal',activation='relu'))
```

```python
model.add(Dense(units=5,kernel_initializer='normal',activation='tanh'))
```

```python
model.add(Dense(1,kernel_initializer='normal'))
```

```python
model.summary()
```

```python
model.compile(loss='mean_squared_error',optimizer='adam')
```

```python
model.fit(x_train,y_train,batch_size=20,epochs=50,verbose=1)
```

```python
model.fit(x_train,y_train,batch_size=20,epochs=32,verbose=0)
```

```python
prediction=model.predict(x_test)
```

```python
prediction=target_scalar.inverse_transform(prediction)
```

```python
y_test=target_scalar.inverse_transform(y_test)
```

```python
test_data = predicted_scalar.inverse_transform(x_test)
```

```python
testing_data=pd.DataFrame(data=test_data)
testing_data.columns = predictor_var
testing_data['Price']=y_test
testing_data['Predicted Price']=prediction[:, 0]
```
